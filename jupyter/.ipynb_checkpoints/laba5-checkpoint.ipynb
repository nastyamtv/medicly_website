```python
import numpy as np
weights = np.array([[0.1, 0.5], [0.2, 0.6]])
X = np.array([1, 0])
D = np.array([0, 1])
learning_rate = 0.8
def sigmoid(x, k=1):
    return 1 / (1 + np.exp(-x * k))
n1_output = sigmoid(np.dot(X, weights[0]))
n2_output = sigmoid(np.dot(X, weights[1]))
errors = D - np.array([n1_output, n2_output])
new_weights = weights - learning_rate * np.outer(errors, X)

print("Y1:", round(n1_output,3))
print("Y2:",round(n2_output,3))
print("E1:",round(errors[0],3))
print("E2:",round(errors[1],3))
print("Оновлені ваги")
print(new_weights[0,1], new_weights[1,1] )
print(round(new_weights[0,0],3), round(new_weights[1,0],3) )
```
