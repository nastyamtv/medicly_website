```python
from PIL import Image
from IPython.display import display

image_path = 'image.jpg'
img = Image.open(image_path)
img = img.resize((300, 600))
display(img)
```

```python
img_gray = img.convert('L')
print("Grayscale Image:")
display(img_gray)
```

```python
# get the width and height of the image
width, height = img.size

# calculate the midpoint of the image
midpoint = width // 2

# split the image into two halves
left_half = img.crop((0, 0, midpoint, height))
right_half = img.crop((midpoint, 0, width, height))

# display the left half
print("Left Half:")
display(left_half)

# display the right half
print("Right Half:")
display(right_half)
```

```python
new_width = img.width * 7  
new_height = img.height
stretched_img = img.resize((new_width, new_height))

display(stretched_img)
```

```python
from PIL import Image, ImageOps
import matplotlib.pyplot as plt
rotated_img = img.rotate(80, expand=True)

mirrored_img = ImageOps.mirror(img)

plt.imshow(rotated_img)
plt.axis('off') 
plt.show()

plt.imshow(mirrored_img)
plt.axis('off') 
plt.show()
```

```python
images = [img, rotated_img]

# Кількість зображень
num_images = len(images)
cols = 2
rows = (num_images + 1) // cols  # Визначення кількості рядків

# Створення фігури
fig, axs = plt.subplots(rows, cols, figsize=(10, 10))

# Відображення зображень у сітці
for i, ax in enumerate(axs.flat):
    if i < num_images:
        ax.imshow(images[i])
        ax.set_title(f'image{i}')
        ax.axis('off')
    else:
        fig.delaxes(ax)  # Видалення порожніх осей, якщо зображень менше, ніж комірок у сітці

# Показати зображення
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('image.jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Resize the image to twice its original length
resized_img1 = cv2.resize(img_rgb, (img_rgb.shape[1] * 3, img_rgb.shape[0]))
resized_img2 = cv2.resize(img_rgb, (img_rgb.shape[1], img_rgb.shape[0]* 3))

fig, axs = plt.subplots(1, 2, figsize=(12, 6))

axs[0].imshow(resized_img1)
axs[0].set_title('Image 1')

axs[1].imshow(resized_img2)
axs[1].set_title('Image 2')

# Layout so plots do not overlap
fig.tight_layout()

plt.show()

```

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

img_gray = cv2.imread('example1.jpg', cv2.IMREAD_GRAYSCALE)
img_gray2 = cv2.imread('example2.jpg', cv2.IMREAD_GRAYSCALE)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(img_gray, cmap='gray')
plt.title('Original Grayscale Image 1')

plt.subplot(1, 2, 2)
plt.imshow(img_gray2, cmap='gray')
plt.title('Original Grayscale Image 2')

plt.show()

threshold_value = 10
threshold_value2 = 200
max_value = 255
ret, img_thresholded = cv2.threshold(img_gray, threshold_value, max_value, cv2.THRESH_BINARY)
ret, img_thresholded2 = cv2.threshold(img_gray2, threshold_value2, max_value, cv2.THRESH_BINARY)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(img_thresholded, cmap='gray')
plt.title('Thresholded Image 1')

plt.subplot(1, 2, 2)
plt.imshow(img_thresholded2, cmap='gray')
plt.title('Thresholded Image 2')

plt.show()

hist = cv2.calcHist([img_gray], [0], None, [256], [0, 256])

plt.figure()
plt.plot(hist)
plt.title('Histogram of Grayscale Image 1')
plt.xlabel('Pixel Value')
plt.ylabel('Frequency')
plt.show()
```

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

img_gray = cv2.imread('example.jpg', cv2.IMREAD_GRAYSCALE)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(img_gray, cmap='gray')
plt.title('1')

img_adaptive_threshold = cv2.adaptiveThreshold(img_gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
                                               cv2.THRESH_BINARY, 11, 2)
plt.subplot(1, 2, 2)
plt.imshow(img_adaptive_threshold, cmap='gray')
plt.title('2')
plt.show()

hist = cv2.calcHist([img_gray], [0], None, [256], [0, 256])

plt.figure()
plt.plot(hist)
plt.title('HIst')
plt.xlabel('Pixel Value')
plt.ylabel('Frequency')
plt.show()
```

```python
import cv2 
import numpy as np 
import matplotlib.pyplot as plt 

# path to input image is specified and 
# image is loaded with imread command 
image1 = cv2.imread('example.jpg') 

# cv2.cvtColor is applied over the 
# image input with applied parameters 
# to convert the image in grayscale 
img = cv2.cvtColor(image1, cv2.COLOR_BGR2GRAY) 

# applying different thresholding 
# techniques on the input image 
# all pixels value above 120 will 
# be set to 255 
ret, thresh1 = cv2.threshold(img, 120, 255, cv2.THRESH_BINARY) 
ret, thresh2 = cv2.threshold(img, 120, 255, cv2.THRESH_BINARY_INV) 
ret, thresh3 = cv2.threshold(img, 120, 255, cv2.THRESH_TRUNC) 
ret, thresh4 = cv2.threshold(img, 120, 255, cv2.THRESH_TOZERO) 
ret, thresh5 = cv2.threshold(img, 120, 255, cv2.THRESH_TOZERO_INV) 

fig, ax = plt.subplots(3, 2, figsize=(12, 20))

ax[0,0].imshow(thresh1, cmap='gray')
ax[0,0].set_title('Binary Threshold')
ax[0,0].axis('off')

ax[0,1].imshow(thresh2, cmap='gray')
ax[0,1].set_title('Binary Inverse Threshold')
ax[0,1].axis('off')

ax[1,0].imshow(thresh3, cmap='gray')
ax[1,0].set_title('Truncate Threshold')
ax[1,0].axis('off')

ax[1,1].imshow(thresh4, cmap='gray')
ax[1,1].set_title('Zero Threshold')
ax[1,1].axis('off')

ax[2,0].imshow(thresh5, cmap='gray')
ax[2,0].set_title('Zero Inverse Threshold')
ax[2,0].axis('off')

plt.show()

# De-allocate any associated memory usage 
cv2.destroyAllWindows()
```
